```python

        # ============================================================
        # CELL 1: INSTALL REQUIRED LIBRARIES (Google Colab/Jupyter)
        # ============================================================

        !pip -q install gradio plotly

        # ============================================================
        # CELL 2: IMPORT LIBRARIES
        # ============================================================

        import pandas as pd
        import numpy as np
        import matplotlib.pyplot as plt
        import seaborn as sns
        import warnings

        from sklearn.preprocessing import StandardScaler
        from sklearn.decomposition import PCA
        from sklearn.cluster import KMeans
        from sklearn.metrics import silhouette_score

        warnings.filterwarnings("ignore")

        print("Libraries imported successfully!")

        # ============================================================
        # CELL 3: LOAD DATASET (WITH BACKUP SOURCE)
        # ============================================================

        url1 = "https://raw.githubusercontent.com/MakroMB/student-performance-analysis/main/data/student-mat.csv"
        url2 = "https://raw.githubusercontent.com/selva86/datasets/master/student-mat.csv"

        try:
            df = pd.read_csv(url1, sep=";")
            print("Dataset loaded successfully from Source 1!")

        except Exception:
            try:
                df = pd.read_csv(url2, sep=";")
                print("Dataset loaded successfully from Backup Source!")
            except Exception as e:
                raise RuntimeError(f"Both dataset sources failed: {e}")

        print("\nDataset Shape:", df.shape)
        display(df.head())

        # ============================================================
        # CELL 4: BASIC DATASET INSPECTION
        # ============================================================

        print("Dataset Shape:", df.shape)

        print("\nColumn Names:")
        print(df.columns.tolist())

        print("\nDataset Information:")
        df.info()

        print("\nMissing Values:")
        print(df.isnull().sum())

        print("\nStatistical Summary:")
        display(df.describe())

        print("\nFirst 5 Rows:")
        display(df.head())

        # ============================================================
        # CELL 5: FINAL GRADE DISTRIBUTION
        # ============================================================

        plt.figure(figsize=(8, 5))

        plt.hist(df["G3"], bins=15, edgecolor="black")

        plt.title("Distribution of Final Grades (G3)")
        plt.xlabel("Final Grade")
        plt.ylabel("Number of Students")
        plt.xticks(range(0, 21, 2))
        plt.grid(axis="y", alpha=0.3)

        plt.show()

        # ============================================================
        # CELL 6: GENDER DISTRIBUTION
        # ============================================================

        gender_counts = df["sex"].value_counts()

        plt.figure(figsize=(6, 4))
        gender_counts.plot(kind="bar", edgecolor="black")

        plt.title("Gender Distribution of Students")
        plt.xlabel("Gender")
        plt.ylabel("Number of Students")
        plt.xticks(rotation=0)
        plt.grid(axis="y", alpha=0.3)

        plt.show()

        print(gender_counts)

        # ============================================================
        # CELL 7: STUDY TIME VS FINAL GRADE
        # ============================================================

        study_grade = df.groupby("studytime")["G3"].mean()

        plt.figure(figsize=(7, 5))
        study_grade.plot(kind="bar", edgecolor="black")

        plt.title("Average Final Grade by Weekly Study Time")
        plt.xlabel("Study Time Category")
        plt.ylabel("Average Final Grade (G3)")
        plt.xticks(rotation=0)
        plt.grid(axis="y", alpha=0.3)

        plt.show()

        print("Average Final Grade by Study Time:")
        display(study_grade.round(2))

        # ============================================================
        # CELL 8: CORRELATION HEATMAP
        # ============================================================

        df_encoded = pd.get_dummies(df, drop_first=True)
        correlation = df_encoded.corr()

        plt.figure(figsize=(16, 12))

        sns.heatmap(
            correlation,
            cmap="coolwarm",
            center=0,
            annot=False,
            linewidths=0.3
        )

        plt.title("Correlation Heatmap of Student Performance Dataset")
        plt.tight_layout()
        plt.show()

        # ============================================================
        # CELL 9: FEATURE SELECTION
        # ============================================================

        feature_cols = [
            "age", "Medu", "Fedu", "traveltime", "studytime", "failures",
            "famrel", "freetime", "goout", "Dalc", "Walc", "health", "absences"
        ]

        work = df[feature_cols + ["G3"]].dropna().copy()

        X = work[feature_cols].copy()
        y_grade = work["G3"].copy()

        print("Rows available for analysis:", len(work))
        print("Features used for clustering:", len(feature_cols))

        display(work.head())

        # ============================================================
        # CELL 10: FEATURE SCALING
        # ============================================================

        scaler = StandardScaler()

        X_scaled = scaler.fit_transform(X)

        print("Feature scaling completed successfully!")
        print("Scaled data shape:", X_scaled.shape)

        display(pd.DataFrame(X_scaled, columns=feature_cols).head())

        # ============================================================
        # CELL 11: PCA DIMENSIONALITY REDUCTION
        # ============================================================

        pca = PCA(n_components=2)

        X_pca = pca.fit_transform(X_scaled)

        print("PCA completed successfully!")
        print("Reduced data shape:", X_pca.shape)

        print("\nExplained Variance Ratio:")
        print(pca.explained_variance_ratio_)

        print("\nTotal Variance Explained:")
        print(round(sum(pca.explained_variance_ratio_) * 100, 2), "%")

        # ============================================================
        # CELL 12: ELBOW METHOD
        # ============================================================

        inertia_values = []
        K_range = range(1, 11)

        for k in K_range:
            model = KMeans(n_clusters=k, random_state=42, n_init=10)
            model.fit(X_scaled)
            inertia_values.append(model.inertia_)

        plt.figure(figsize=(8, 5))
        plt.plot(K_range, inertia_values, marker="o")

        plt.title("Elbow Method for Optimal K")
        plt.xlabel("Number of Clusters (K)")
        plt.ylabel("Inertia (WCSS)")
        plt.xticks(K_range)
        plt.grid(alpha=0.3)

        plt.show()

        # ============================================================
        # CELL 13: K-MEANS CLUSTERING
        # ============================================================

        kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

        clusters = kmeans.fit_predict(X_scaled)

        work["Cluster"] = clusters

        print("K-Means clustering completed successfully!")

        print("\nCluster Distribution:")
        print(work["Cluster"].value_counts().sort_index())

        # ============================================================
        # CELL 14: PCA CLUSTER VISUALIZATION
        # ============================================================

        plt.figure(figsize=(9, 6))

        plt.scatter(
            X_pca[:, 0],
            X_pca[:, 1],
            c=clusters,
            cmap="viridis",
            s=50,
            alpha=0.7
        )

        plt.title("Student Clusters Visualized using PCA")
        plt.xlabel("Principal Component 1")
        plt.ylabel("Principal Component 2")

        plt.colorbar(label="Cluster")
        plt.grid(alpha=0.3)
        plt.show()

        # ============================================================
        # CELL 15: SILHOUETTE SCORE
        # ============================================================

        score = silhouette_score(X_scaled, clusters)

        print("Silhouette Score:", round(score, 3))

        # ============================================================
        # CELL 16: CLUSTER-WISE ACADEMIC PERFORMANCE
        # ============================================================

        cluster_performance = work.groupby("Cluster")["G3"].agg(
            ["count", "mean", "min", "max"]
        )

        cluster_performance.columns = [
            "Number of Students",
            "Average Final Grade",
            "Minimum Grade",
            "Maximum Grade"
        ]

        print("Academic Performance Across Clusters:")
        display(cluster_performance.round(2))

        plt.figure(figsize=(8, 5))

        cluster_performance["Average Final Grade"].plot(
            kind="bar",
            edgecolor="black"
        )

        plt.title("Average Final Grade Across Student Clusters")
        plt.xlabel("Cluster")
        plt.ylabel("Average Final Grade (G3)")
        plt.xticks(rotation=0)
        plt.grid(axis="y", alpha=0.3)

        plt.show()

        # ============================================================
        # CELL 17: CLUSTER CHARACTERISTICS
        # ============================================================

        cluster_profiles = work.groupby("Cluster")[feature_cols].mean()

        print("Cluster-wise Student Characteristics:")
        display(cluster_profiles.round(2))

        plt.figure(figsize=(14, 7))

        sns.heatmap(
            cluster_profiles,
            annot=True,
            cmap="YlGnBu",
            fmt=".2f",
            linewidths=0.5
        )

        plt.title("Comparison of Student Characteristics Across Clusters")
        plt.xlabel("Student Features")
        plt.ylabel("Cluster")

        plt.tight_layout()
        plt.show()

        # ============================================================
        # CELL 18: FINAL GRADE DISTRIBUTION BY CLUSTER
        # ============================================================

        plt.figure(figsize=(9, 6))

        sns.boxplot(
            data=work,
            x="Cluster",
            y="G3"
        )

        plt.title("Final Grade Distribution by Student Clusters")
        plt.xlabel("Cluster")
        plt.ylabel("Final Grade (G3)")
        plt.grid(axis="y", alpha=0.3)

        plt.show()

        # ============================================================
        # CELL 19: INTERACTIVE GRADIO DASHBOARD
        # ============================================================

        import gradio as gr
        import plotly.express as px

        def make_dashboard(k=3):

            k = int(k)

            # PCA for visualization
            pca_model = PCA(n_components=2)
            pcs = pca_model.fit_transform(X_scaled)

            # K-Means on original scaled features (consistent with main model)
            model = KMeans(n_clusters=k, random_state=42, n_init=10)
            labels = model.fit_predict(X_scaled)

            results = work.copy()
            results["PC1"] = pcs[:, 0]
            results["PC2"] = pcs[:, 1]
            results["Cluster"] = labels.astype(str)

            # Silhouette score
            dashboard_score = silhouette_score(X_scaled, labels)

            # PCA Scatter Plot
            scatter = px.scatter(
                results,
                x="PC1",
                y="PC2",
                color="Cluster",
                hover_data=["G3", "studytime", "failures", "absences"],
                title=f"Student Clusters (K={k})"
            )

            # Grade Distribution Plot
            grade_plot = px.box(
                results,
                x="Cluster",
                y="G3",
                color="Cluster",
                title="Academic Performance Across Clusters"
            )

            # Cluster Profile Table
            profile = results.groupby("Cluster")[feature_cols + ["G3"]].mean().round(2)
            profile.insert(
                0,
                "Number of Students",
                results.groupby("Cluster").size()
            )

            summary = f"""
            ## Dashboard Summary

            - Total Students: {len(results)}
            - Features Used: {len(feature_cols)}
            - Number of Clusters: {k}
            - PCA Variance Explained: {pca_model.explained_variance_ratio_.sum()*100:.2f}%
            - Silhouette Score: {dashboard_score:.3f}
            """

            return summary, scatter, grade_plot, profile.reset_index()


        with gr.Blocks(theme=gr.themes.Soft()) as demo:

            gr.Markdown("# Student Performance Analysis Dashboard")
            gr.Markdown("Explore student patterns using PCA and K-Means Clustering.")

            k_slider = gr.Slider(
                minimum=2,
                maximum=6,
                value=3,
                step=1,
                label="Select Number of Clusters"
            )

            output_summary = gr.Markdown()
            output_scatter = gr.Plot(label="PCA Cluster Visualization")
            output_grades = gr.Plot(label="Academic Performance")
            output_profile = gr.Dataframe(label="Cluster Profiles")

            update_btn = gr.Button("Analyze Students")

            update_btn.click(
                fn=make_dashboard,
                inputs=k_slider,
                outputs=[
                    output_summary,
                    output_scatter,
                    output_grades,
                    output_profile
                ]
            )

            demo.load(
                fn=make_dashboard,
                inputs=k_slider,
                outputs=[
                    output_summary,
                    output_scatter,
                    output_grades,
                    output_profile
                ]
            )

        demo.launch(share=True)

        # ============================================================
        # CELL 20: PROJECT CONCLUSION
        # ============================================================

        print("=" * 60)
        print("       STUDENT PERFORMANCE ANALYSIS - CONCLUSION")
        print("=" * 60)

        print("\n1. DATASET OVERVIEW")
        print(f"Total Students Analyzed: {len(work)}")
        print(f"Features Used: {len(feature_cols)}")
        print("Machine Learning Algorithm: K-Means Clustering")

        print("\n2. PCA ANALYSIS")
        print(
            f"Total Variance Explained: "
            f"{pca.explained_variance_ratio_.sum() * 100:.2f}%"
        )

        print("\n3. CLUSTERING RESULTS")
        print(f"Number of Clusters: {work['Cluster'].nunique()}")
        print(f"Silhouette Score: {score:.3f}")

        print("\n4. ACADEMIC PERFORMANCE")
        display(cluster_performance.round(2))

        best_cluster = cluster_performance["Average Final Grade"].idxmax()

        print(
            f"\nCluster {best_cluster} has the highest average final grade: "
            f"{cluster_performance.loc[best_cluster, 'Average Final Grade']:.2f}"
        )

        print("\n5. FINAL CONCLUSION")
        print(
            "PCA and K-Means clustering identified groups of students "
            "based on selected demographic, study and lifestyle characteristics."
        )

        print(
            "The analysis helps explore patterns between student characteristics "
            "and academic performance. Clustering shows associations, "
            "not direct cause-and-effect relationships."
        )

        print("\n" + "=" * 60)
        print("                 PROJECT COMPLETED")
        print("=" * 60)

        # ============================================================
        # CELL 21: EXPORT FINAL RESULTS
        # ============================================================

        output_file = "Student_Performance_Clustering_Results.csv"

        work.to_csv(output_file, index=False)

        print("Results exported successfully!")
        print("File Name:", output_file)
        print("Total Records Exported:", len(work))

        # ============================================================
        # CELL 22: FINAL PROJECT VALIDATION
        # ============================================================

        import os

        print("=" * 50)
        print("         FINAL PROJECT VALIDATION")
        print("=" * 50)

        print("\n1. Dataset Loaded:", not df.empty)
        print("2. Scaled Data Shape:", X_scaled.shape)
        print("3. PCA Output Shape:", X_pca.shape)
        print("4. Total Clusters:", work["Cluster"].nunique())
        print("5. Missing Values:", work.isnull().sum().sum())
        print("6. CSV Exported:", os.path.exists(output_file))

        print("\nCluster Distribution:")
        print(work["Cluster"].value_counts().sort_index())

        print("\nAll validation checks completed!")
        print("=" * 50)
```
